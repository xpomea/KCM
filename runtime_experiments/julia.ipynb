```julia
using Random
using Printf

function matrix_multiply(n)
    # Generate two random n x n matrices
    A = rand(1:10, n, n)  # Matrix A with random integers between 1 and 10
    B = rand(1:10, n, n)  # Matrix B with random integers between 1 and 10

    # Initialize the result matrix C with zeros
    C = zeros(Int, n, n)   # Result matrix initialized to 0

    # Start measuring time
    start_time = time()

    # Perform matrix multiplication: C = A * B
    for i in 1:n
        for j in 1:n
            for k in 1:n
                C[i, j] += A[i, k] * B[k, j]
            end
        end
    end

    # End measuring time
    end_time = time()
    println("Matrix multiplication took $(end_time - start_time) seconds.")
end

# Set matrix size
n = 1000

# Call the function
matrix_multiply(n)
```
