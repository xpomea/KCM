```python
import numpy as np
import time

n = 20
total_time_ms = 0.0

for i in range(n):
    arr = np.random.rand(int(1e5))
    start = time.time()
    sorted_arr = np.sort(arr)
    end = time.time()
    iter_time_ms = (end - start) * 1000  # перевод в миллисекунды
    total_time_ms += iter_time_ms
    print(f"#{i+1}: {iter_time_ms:.3f} ms")

mean_time = total_time_ms / n
print(f"Mean iteration time: {mean_time:.3f} ms")
```

```python
import random
import time

n = 20
m = int(1e5)
total_time_ms = 0.0

for i in range(n):
    arr = [random.random() for _ in range(m)]
    start = time.time()
    sorted_arr = sorted(arr)
    end = time.time()
    iter_time_ms = (end - start) * 1000
    total_time_ms += iter_time_ms
    print(f"#{i+1}: {iter_time_ms:.3f} ms")

mean_time = total_time_ms / n
print(f"Mean iteration time: {mean_time:.3f} ms")
```

```python
import random
import time

def naive_matrix_multiply(n):
    # Create two random n x n matrices
    A = [[random.randint(1, 10) for _ in range(n)] for _ in range(n)]
    B = [[random.randint(1, 10) for _ in range(n)] for _ in range(n)]

    # Initialize the result matrix C with zeros
    C = [[0 for _ in range(n)] for _ in range(n)]

    # Start measuring time
    start_time = time.time()

    # Perform matrix multiplication: C = A * B
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]

    # End measuring time
    end_time = time.time()
    print(f"Matrix multiplication took {end_time - start_time} seconds.")

# Set matrix size
n = 1000

# Call the function
naive_matrix_multiply(n)
```

```python
import numpy as np
import time

def optimized_matrix_multiply(n):
    # Create two random n x n matrices using NumPy
    A = np.random.randint(1, 11, size=(n, n))  # Random integers between 1 and 10
    B = np.random.randint(1, 11, size=(n, n))

    # Start measuring time
    start_time = time.time()

    # Perform optimized matrix multiplication using np.dot
    C = np.dot(A, B)

    # End measuring time
    end_time = time.time()
    print(f"Optimized matrix multiplication took {end_time - start_time} seconds.")

# Set matrix size
n = 1000

# Call the function
optimized_matrix_multiply(n)
```
